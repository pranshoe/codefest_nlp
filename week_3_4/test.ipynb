Load Prebuilt Model

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("pranshoe/shakespeare-google-t5")
model = AutoModelForSeq2SeqLM.from_pretrained("pranshoe/shakespeare-google-t5")
```

Define Model Function which takes an input text and outputs the translated text

```python
def speak_shakespeare(text):
    input_text = "translate modern English to Shakespearean English: " + text
    inputs = tokenizer(input_text, return_tensors="pt").to(model.device)

    outputs = model.generate(
        **inputs,
        max_length=128,
        num_beams=5,
        early_stopping=True
    )

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

Create Gradio UI and Start it

```python
import gradio as gr

with gr.Blocks() as demo:
    gr.Markdown("# Modern English to Shakespearean Language")
    with gr.Row(equal_height=True):
        input = gr.Textbox(show_label=False, placeholder="Enter you text", scale=4)
        button = gr.Button("Translate", scale=1)
    
    output=gr.Textbox(label="Output")

    input.submit(
        fn=speak_shakespeare,
        inputs=input,
        outputs=output
    )

    button.click(
        fn=speak_shakespeare,
        inputs=input,
        outputs=output
    )

demo.launch()
```
